#### Here is the box and whisker plot of all the features. As you can see, there are outliers that fall outside of the normal ranges.

```python
features = ['X1', 'X2', 'X3', 'X4', 'X5', 'X6']

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df[features],
    palette='Set2'
)

plt.title('Box Plots of Features')
plt.ylabel('Y')
plt.show()
```

#### We replaced the outliers for feature X2 with the median.

```python

Q1 = df['X2'].quantile(0.25)
Q3 = df['X2'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

median = df['X2'].median()

df.loc[df['X2'] < lower_bound, 'X2'] = median
df.loc[df['X2'] > upper_bound, 'X2'] = median
    
```

#### We replaced the outliers for feature X3 with the median.

#### We replaced the outliers for feature X4 with the median.
